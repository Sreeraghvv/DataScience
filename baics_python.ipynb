### Exercise 1: Prime Numbers

```python
num = int(input("Enter a number: "))
flag = False

if num == 0 or num == 1:
    print(num, "is not a prime number")
elif num > 1:
    for i in range(2, num):
        if num % i == 0:
            flag = True
            break

    if flag:
        print(num, "is not a prime number")
    else:
        print(num, "is a prime number")
```

### Exercise 2: Product of Random Numbers

```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print(f"What is the product of {num1} and {num2}?")
user_answer = int(input("Enter your answer: "))

if user_answer == num1 * num2:
    print("Correct")
else:
    print(f"Incorrect. The correct answer is {num1 * num2}.")
```

### Exercise 3: Squares of Even/Odd Numbers

```python
# This is the code to print square of all even numbers between 100 and 200
for num in range(100, 201):
    if num % 2 == 0:
        print(f"{num}² = {num ** 2}")
```

### Exercise 4:Word counter

```python
from collections import Counter

input_text = input('Enter the sentence:')

# Convert to lowercase and split into words
words = input_text.lower().split()

# Count words directly
word_count = Counter(words)

# Print the result
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

## Exercise 5: Check for Palindrome

```python
# Take input from the user
text = input("Enter a string: ")
# Convert to lowercase
text = text.lower()
# Remove spaces and punctuation
text = ''.join(char for char in text if char.isalnum())
# Check if it reads the same forward and backward
if text == text[::-1]:
    print(text,"is a palindrome ")
else:
    print(text,"is not a palindrome ")
```
