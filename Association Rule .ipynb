```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
```

```python
df = pd.read_excel("Online retail.xlsx")
df.head()
```

```python
df.info()
```

```python
# Identify missing value
df.isnull().sum()
```

```python
# Extract the first column (transactions)
# Your dataset likely has one column where each row = one transaction
transactions = df.iloc[:, 0].dropna().apply(lambda x: [item.strip() for item in str(x).split(',')])
```

```python
transactions
```

```python
#Use TransactionEncoder to convert into 1/0 (Boolean) format
te = TransactionEncoder()
enc_matrix = te.fit(transactions).transform(transactions)
```

```python
te.columns_
```

```python
encoded_df = pd.DataFrame(enc_matrix, columns=te.columns_)
encoded_df
```

```python
#  Apply Apriori Algorithm
frequent_items = apriori(encoded_df, min_support=0.02, use_colnames=True)
frequent_items
```

```python
#  Generate Association Rules
rules = association_rules(frequent_items, metric="lift", min_threshold=1.5)
rules
```

```python
rules[rules['lift']>1.5]
```

#### Analysis and Interpretation:

###### From the generated association rules, it is observed that products like **mineral water, spaghetti, milk, and ground beef** frequently appear together, indicating strong purchasing relationships. Rules with higher **lift values (above 1.5)** and moderate **support (above 0.02)** suggest meaningful associations — for example, customers who buy spaghetti are about **2.3 times more likely** to buy ground beef. Such insights can help in **cross-selling strategies**, optimizing **store layouts**, and improving **inventory management** by ensuring complementary products are stocked together. Overall, association rule mining provides valuable understanding of **customer buying patterns**, enabling data-driven marketing and sales decisions.


#### Lift and its Importance:

###### Lift measures how much more likely two items are to be bought together compared to being bought independently. It is calculated as the ratio of the observed support to the expected support if the items were independent. A lift value greater than 1 indicates a positive association, meaning the items are frequently purchased together. It is important because it helps identify strong, non-random relationships between products, providing deeper insights than confidence alone.

#### Support and Confidence:

###### Support indicates how often a particular itemset appears in the dataset, showing the overall frequency of a rule. It is calculated as the number of transactions containing the itemset divided by the total number of transactions. Confidence measures the reliability of a rule, calculated as the ratio of the number of transactions containing both items (A and B) to the number containing only A. Together, they help evaluate the strength and significance of association rules.

#### Limitations of Association Rule Mining:

###### Some key challenges include generating a large number of irrelevant or redundant rules, especially in large datasets. Setting proper thresholds for support and confidence is difficult — too high may miss useful patterns, while too low may produce too many weak rules. Additionally, association rules only capture co-occurrence relationships and not causality, which can sometimes lead to misleading conclusions.
